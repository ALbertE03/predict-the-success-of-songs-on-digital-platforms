```python
df1['(x?)'].fillna(1, inplace=True)
```

```python
def aux(x):
    try:
        return int(x)
    except:
        return int(str(x).replace("(x",'').replace(')','').strip())
    

df1['(x?)'] = df1['(x?)'].apply(aux)
```

```python
df1
```

```python
def clean_number(x):
    return int(str(x).replace(",", '').strip())

df1['PkStreams'] = df1['PkStreams'].apply(clean_number)
df1['Total'] = df1['Total'].apply(clean_number)
```

```python
df1['artist'] = df1['Artist and Title'].apply(lambda x: x.split('-')[0])
df1['songs'] = df1['Artist and Title'].apply(lambda x: ' '.join(x.split('-')[1:]))
```
