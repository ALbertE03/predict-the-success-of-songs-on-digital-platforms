```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

k_range = range(1, 11)
inertias = []

for k in k_range:
    kmeans = KMeans(n_clusters=k, n_init='auto', random_state=42)
    kmeans.fit(X_scaled)
    inertias.append(kmeans.inertia_)  


plt.figure(figsize=(10, 6))
plt.plot(k_range, inertias, 'bo-', markersize=8, linewidth=2)
plt.xlabel('Número de Clusters (k)', fontsize=12)
plt.ylabel('Inercia', fontsize=12)
plt.title('Método del Codo para K-Means', fontsize=14)
plt.xticks(k_range)

plt.legend()
plt.grid(True)
plt.show()
```

```python

optimal_k = 5  
kmeans = KMeans(n_clusters=optimal_k, n_init='auto')
df['cluster'] = kmeans.fit_predict(X_scaled)


cluster_stats = df.groupby('cluster').agg({
    'log_streams': ['mean', 'count'],
    'recency_score': 'mean',
    'danceability': 'mean'
}).round(2)

print("Estadísticas por Cluster:")
cluster_stats
```

```python
import umap.umap_ as umap
import matplotlib.pyplot as plt
import seaborn as sns


reducer = umap.UMAP()

X_umap = reducer.fit_transform(X_scaled)

plt.figure(figsize=(12, 8))
scatter = plt.scatter(
    X_umap[:, 0], 
    X_umap[:, 1], 
    c=df['cluster'], 
    cmap='Spectral',
    s=10,             
    alpha=0.8,        
    edgecolor='k',   
    linewidth=0.3   
)

centers_umap = reducer.transform(kmeans.cluster_centers_)
plt.scatter(
    centers_umap[:, 0], 
    centers_umap[:, 1], 
    c='black', 
    marker='X', 
    s=200, 
    label='Centroides'
)


plt.colorbar(scatter, label='Cluster', pad=0.01)
plt.title('Visualización de Clusters con UMAP (k=5)', fontsize=16, pad=20)
plt.xlabel('Dimensión UMAP 1', fontsize=12)
plt.ylabel('Dimensión UMAP 2', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.3)


plt.legend(fontsize=10)
plt.tight_layout()

print("\nVarianza preservada (estimada): ~{:.1f}%".format(
    reducer.embedding_[0].var() / X_scaled.var() * 100
))
ch_score = calinski_harabasz_score(X_scaled, df['cluster'])
db_score = davies_bouldin_score(X_scaled, df['cluster'])
print(ch_score)
print(db_score)
s = silhouette_score(X_scaled,df['cluster'])
print(s)
plt.show()
```
